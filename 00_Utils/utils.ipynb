```python
from pyspark.sql import functions as F
```

```python
"""Função simples que posso passar um df completo, e quais colunas eu quero verificar se existem dados duplicados"""
def verificar_duplicidade(df, colunas):
    duplicados = (
        df.groupBy(*colunas)
          .count()
          .filter(F.col("count") > 1)
    )

    qtd = duplicados.count()

    if qtd == 0:
        print(f"Nenhum dado duplicado {colunas}")
    else:
        print(f"⚠️ {qtd} chaves duplicadas encontradas: {colunas}")
        duplicados.show(truncate=False)
```
