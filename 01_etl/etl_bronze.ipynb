```python
# Vamos listar arquivos do volume
volume_path = "/Volumes/workspace/lakehouse_panvel/volume_bronze/"

files = dbutils.fs.ls(volume_path)

# Exibir informações dos arquivos
for file in files:
    print(f"Nome: {file.name}")
    print(f"Caminho: {file.path}")
```

### Categorias

```python
df_categorias = (
    spark.read
        .option("multiline", "true")
        .json("/Volumes/workspace/lakehouse_panvel/volume_bronze/categorias.json")
)

df_categorias.show(truncate=False)
```

```python
df_categorias.write.mode("overwrite").format("delta").saveAsTable("workspace.lakehouse_panvel.bronze_categorias")
```

### Filiais

```python
df_filiais = spark.read.parquet("/Volumes/workspace/lakehouse_panvel/volume_bronze/filiais.parquet")

df_filiais.show(truncate=False)
```

```python
df_filiais.write.mode("overwrite").format("delta").saveAsTable("workspace.lakehouse_panvel.bronze_filiais")
```

### Produtos

```python
df_produtos = spark.read.csv("/Volumes/workspace/lakehouse_panvel/volume_bronze/produtos.csv", header=True, inferSchema=True)

df_produtos.show(truncate=False)
```

```python
df_produtos.write.mode("overwrite").format("delta").saveAsTable("workspace.lakehouse_panvel.bronze_produtos")
```

### Vendas do Varejo

```python
df_vendas = spark.read.parquet("/Volumes/workspace/lakehouse_panvel/volume_bronze/vendas_varejo.parquet")

df_vendas.show(truncate=False)
```

```python
df_vendas.write.mode("overwrite").format("delta").saveAsTable("workspace.lakehouse_panvel.bronze_vendas")
```
